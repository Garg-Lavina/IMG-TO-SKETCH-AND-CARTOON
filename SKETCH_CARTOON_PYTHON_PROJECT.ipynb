This code uses OpenCV to read an input image, process it, and display three versions: the original (resized), an edge-detected mask, and a cartoonized effect. It first converts the image to grayscale, applies noise reduction and adaptive thresholding to detect edges, then smooths the color image with a bilateral filter. Finally, it combines the smoothed colors and edge mask to create a cartoon-like version of the image, displaying all results in separate windows for comparison.

```python
import cv2                    # OpenCV library for image processing
#import numpy as np            # NumPy library for numerical operations (not directly used here but often useful)

# Step 1: Read the input image from the specified path
img = cv2.imread("final.jpeg")
# Step 2: Resize the original image to half of its original size for easier display and processing
img_half = cv2.resize(img, (0, 0), fx = 0.5, fy = 0.5) # Edges
# Step 3: Convert the original image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# Step 4: Apply a median blur to the grayscale image to reduce noise- SMOOTHEN OUT IMG
gray = cv2.medianBlur(gray, 1)
# Step 5: Apply adaptive thresholding to detect edges in the image
edges = cv2.adaptiveThreshold(
    gray,                        # Input grayscale image

    255,                         # Maximum value assigned to white pixels

    cv2.ADAPTIVE_THRESH_MEAN_C,  # Adaptive method: mean of neighbourhood

    cv2.THRESH_BINARY,           # Type of thresholding (binary)

    blockSize=11,                # Size of the neighbourhood area

    C=11                         # Constant subtracted from mean to fine-tune threshold

)
# Step 6: Resize the edge-detected image to half for display
e_half = cv2.resize(edges, (0, 0), fx=0.5, fy=0.5)
# Cartoonization
# Step 7: Apply bilateral filter to smooth the image while preserving edges
color = cv2.bilateralFilter(img, d=9, sigmaColor=250, sigmaSpace=250)
# Step 8: Combine the smoothed image and edge mask to create cartoon effect
cartoon = cv2.bitwise_and(color, color, mask=edges)
# Step 9: Resize cartoon image to half for display
half = cv2.resize(cartoon, (0, 0), fx=0.5, fy=0.5)
# Step 10: Show all processed images in separate windows
cv2.imshow("Original Image (Half Size)", img_half)
cv2.imshow("Edge Mask (Half Size)", e_half)
cv2.imshow("Cartoonized Image (Half Size)", half)
# Step 11: Wait for user to press any key, then close all image windows
cv2.waitKey(0)
cv2.destroyAllWindows()
```
